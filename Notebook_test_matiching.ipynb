```python

import pandas
def generate_html_report(report_df, out_path="rapport.html"):
    """Génère un rapport HTML avec un code couleur par statut (vert/jaune/rouge),
    sans dépendre de jinja2."""
    couleurs = {
        "trouvé": "#c8f7c5",
        "plusieurs candidats": "#fff3b0",
        "aucune correspondance": "#f7c5c5",
    }

    lignes_html = []
    for _, row in report_df.iterrows():
        color = couleurs.get(row["Statut"], "white")
        cellules = "".join(f"<td>{row[col]}</td>" for col in report_df.columns)
        lignes_html.append(f'<tr style="background-color: {color};">{cellules}</tr>')

    entetes = "".join(f"<th>{col}</th>" for col in report_df.columns)

    html = f"""
    <html>
    <head>
        <meta charset="utf-8">
        <style>
            table {{ border-collapse: collapse; width: 100%; font-family: sans-serif; }}
            th, td {{ border: 1px solid #ccc; padding: 6px 10px; text-align: left; }}
            th {{ background-color: #333; color: white; }}
        </style>
    </head>
    <body>
        <h2>Rapport de matching BOM / Bibliothèque</h2>
        <table>
            <tr>{entetes}</tr>
            {"".join(lignes_html)}
        </table>
    </body>
    </html>
    """

    with open(out_path, "w", encoding="utf-8") as f:
        f.write(html)
    print(f"Rapport généré : {out_path}")


if __name__ == "__main__":
    library = pd.read_excel("lib/Liste des composants.xlsx")
    BOM = clean_BOM("data/Carte_radar_V0.2--BoM.xlsx")

    report_df = build_report_rows(BOM, library)

    n = len(report_df)
    i = (report_df["Statut"] == "trouvé").sum()
    j = (report_df["Statut"] == "plusieurs candidats").sum()
    print(f"Nombre de composants trouvés : {i} sur {n} avec {j} lignes qui ont généré plusieurs résultats")

    generate_html_report(report_df, "output/rapport.html")
```
